```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# Cargar datos 
X_train_hta = pd.read_csv("../data/processed/X_train_hta_df.csv")
X_test_hta = pd.read_csv("../data/processed/X_test_hta_df.csv")
y_train_hta = pd.read_csv("../data/processed/y_train_hta.csv")
y_test_hta = pd.read_csv("../data/processed/y_test_hta.csv")

sample_weight_train_hta = pd.read_csv("../data/processed/sample_weight_train_hta.csv")

print("Datos listos para empezar a entrenar modelos")
```

# ENTRENAMIENTO CON REGRESIÓN LOGÍSTICA

```python
# ===================================================================
# ENTRENAMIENTO CON CLASS_WEIGHT Y FACTOR ADULTO CON AJUSTE DE UMBRAL 
# ===================================================================

reg_log_hta_class_weight = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Entrenar modelo
reg_log_hta_class_weight.fit(X_train_hta, y_train_hta.values.ravel())

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' ENTRENADA")
```

```python
# ======================================================
# ENTRENAMIENTO CON PESOS BALANCEADOS Y FACTOR ADULTO
# ======================================================

reg_log_hta = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Mismo código, pero añadiendo el sample_weight en el fit:
reg_log_hta.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel(),)

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' Y 'sample_weight' ENTRENADA")
```

```python
# ========================
# ENTRENAMIENTO CON SMOTE 
# ========================

from imblearn.over_sampling import SMOTE

# Se aplica SMOTE únicamente al conjunto de entrenamiento 
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel()
)

# Se entrena la Regresión Logística con los datos sintéticos 
reg_log_hta_smote = LogisticRegression(max_iter=1000, random_state=42)
reg_log_hta_smote.fit(X_train_smote, y_train_smote)

print("MODELO REGRESIÓN LOGÍSTICA CON 'SMOTE' ENTRENADA")
```

```python
# =====================================
# GUARDAR MODELOS REGRESIÓN LOGÍSTICA 
# =====================================

import joblib

# Guardar modelo con 'class_weight'
joblib.dump(reg_log_hta_class_weight, "../models/reg_log_hta_class_weight.joblib")

# Guardar modelo con 'class_weight' y 'sample_weight' 
joblib.dump(reg_log_hta, "../models/reg_log_hta.joblib")

# Guardar modelo con SMOTE 
joblib.dump(reg_log_hta_smote, "../models/reg_log_hta_smote.joblib")

print("Modelos de Regresión Logística guardados con éxito")
```

# ENTRENAMIENTO CON RANDOM FOREST

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' HIPERTENSIÓN 
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf_hta_class_weight = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_hta_class_weight.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_hta_class_weight, "../models/random_forest_hta_class_weight.joblib")
```

```python
# ==============================================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' Y 'sample_weight' HIPERTENSIÓN 
# ==============================================================================

rf_hta = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_hta.fit(X_train_hta, y_train_hta.values.ravel(), 
           sample_weight=sample_weight_train_hta.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_hta, "../models/random_forest_hta.joblib")
```

```python
# ======================================
# ENTRENAMIENTO RANDOM FOREST CON SMOTE 
# ======================================

smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())

print(f"Filas originales en train: {len(X_train_hta)} | Filas tras SMOTE: {len(X_train_smote)}")

rf_smote_hta = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1)

rf_smote_hta.fit(X_train_smote, y_train_smote)

print("MODELO RANDOM FOREST CON SMOTE ENTRENADO CORRECTAMENTE")

# Guardar modelo 
joblib.dump(rf_smote_hta, "../models/rf_smote_hta.joblib")
```

```python
# ========================================
# MODELO RANDOM FOREST C/ PESOS DINÁMICOS 
# ========================================

rf_subsample_hta = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample", # Pesos dinamicos por arbol 
    max_depth=10,  # controlar overfiting
    random_state=42,
    n_jobs=-1
)

rf_subsample_hta.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest con ajuste dinamico por árbol entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_subsample_hta, "../models/subsample_random_forest_hta.joblib")
```

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON Balanced Random Forest
# ============================================================

from imblearn.ensemble import BalancedRandomForestClassifier
brf_hta = BalancedRandomForestClassifier(
    n_estimators=300,
    sampling_strategy="all", # Equilibrar las clases en cada arbol
    replacement=True,
    random_state=42,
    n_jobs=-1
)

brf_hta.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest mejorado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_hta, "../models/balanced_random_forest_hta.joblib")
```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest optimizado
# ===============================================


brf_opt = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=12, 
    min_samples_leaf=5,  
    sampling_strategy="auto",
    random_state=42,
    n_jobs=-1)

# Entrenar incorporando el peso del diseño muestral 
brf_opt.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel())


print("Modelo Balanced Random Forest optimizado")

# Guardar modelo 
joblib.dump(brf_opt, "../models/brf_hta_opt.joblib")
```

# SVM

```python
# =========================================
# MODELO SVM 'class_weight' - HIPERTENSIÓN 
# =========================================

from sklearn.svm import SVC

svm_hta_cw = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

# Entrenamiento 
svm_hta_cw.fit(X_train_hta, y_train_hta.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_hta_cw, "../models/svm_hta_cw.joblib")
```

```python
# ===========================================================
# MODELO SVM 'class_weight' + 'sample_weight' - HIPERTENSIÓN 
# ===========================================================


svm_hta = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_hta.fit(X_train_hta, y_train_hta.values.ravel(), 
           sample_weight=sample_weight_train_hta.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_hta, "../models/svm_hta.joblib")
```

```python
# ==============================
# SVM CON SMOTE - HIPERTENSIÓN 
# ==============================
svm_smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_hta_smote, y_train_hta_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())

svm_hta_smote = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_hta_smote.fit(X_train_hta_smote, y_train_hta_smote)

print("SVM con SMOTE entrenado correctamente")

#Guardar modelo 
joblib.dump(svm_hta_smote, "../models/svm_hta_smote")
```

# MODELOS XGBOOST

```python
# ================================================
# MODELO XGBOOST - 'class_weight' - HIPERTENSIÓN 
# ================================================

from xgboost import XGBClassifier

# Gestión del desbalance equivalente a 'class_weight' 
y_train_limpio = y_train_hta.values.ravel()
equivalente_class_weight = float((y_train_limpio == 0).sum() / (y_train_limpio == 1).sum())

xgb_model_cw_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_cw_hta.fit(X_train_hta, y_train_limpio)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_cw_hta, "../models/xgb_model_cw_hta.joblib")
```

```python
# ==============================================================
# MODELO XGBOOST - 'class_weight' y FACTORADULTO - HIPERTENSIÓN 
# ==============================================================

xgb_model_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_hta.fit(X_train_hta, y_train_limpio, 
                  sample_weight=sample_weight_train_hta.values.ravel())

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_hta, "../models/xgb_model_hta.joblib")
```

```python
# ========================================
# MODELO XGBOOST - 'SMOTE' - HIPERTENSIÓN 
# ========================================

smote = SMOTE(random_state=42)
x_train_smote_xg, y_train_smote_xg = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())


xgb_model_smote_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    random_state = 42,
    n_jobs = -1)

xgb_model_smote_hta.fit(x_train_smote_xg, y_train_smote_xg)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_smote_hta, "../models/xgb_model_smote_hta.joblib")
```
