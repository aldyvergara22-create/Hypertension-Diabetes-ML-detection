```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# Cargar datos 
X_train_hta = pd.read_csv("../data/processed/X_train_hta_df.csv")
X_test_hta = pd.read_csv("../data/processed/X_test_hta_df.csv")
y_train_hta = pd.read_csv("../data/processed/y_train_hta.csv")
y_test_hta = pd.read_csv("../data/processed/y_test_hta.csv")

sample_weight_train_hta = pd.read_csv("../data/processed/sample_weight_train_hta.csv")

print("Datos listos para empezar a entrenar modelos")
```

# ENTRENAMIENTO CON REGRESIÓN LOGÍSTICA

```python
# ===================================================================
# ENTRENAMIENTO CON CLASS_WEIGHT Y FACTOR ADULTO CON AJUSTE DE UMBRAL 
# ===================================================================

reg_log_hta_class_weight = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Entrenar modelo
reg_log_hta_class_weight.fit(X_train_hta, y_train_hta.values.ravel())

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' ENTRENADA")
```

```python
# ======================================================
# ENTRENAMIENTO CON PESOS BALANCEADOS Y FACTOR ADULTO
# ======================================================

reg_log_hta = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Mismo código, pero añadiendo el sample_weight en el fit:
reg_log_hta.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel(),)

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' Y 'sample_weight' ENTRENADA")
```

```python
# ========================
# ENTRENAMIENTO CON SMOTE 
# ========================

from imblearn.over_sampling import SMOTE

# Se aplica SMOTE únicamente al conjunto de entrenamiento 
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel()
)

# Se entrena la Regresión Logística con los datos sintéticos 
reg_log_hta_smote = LogisticRegression(max_iter=1000, random_state=42)
reg_log_hta_smote.fit(X_train_smote, y_train_smote)

print("MODELO REGRESIÓN LOGÍSTICA CON 'SMOTE' ENTRENADA")
```

```python
# =====================================
# GUARDAR MODELOS REGRESIÓN LOGÍSTICA 
# =====================================

import joblib

# Guardar modelo con 'class_weight'
joblib.dump(reg_log_hta_class_weight, "../models/reg_log_hta_class_weight.joblib")

# Guardar modelo con 'class_weight' y 'sample_weight' 
joblib.dump(reg_log_hta, "../models/reg_log_hta.joblib")

# Guardar modelo con SMOTE 
joblib.dump(reg_log_hta_smote, "../models/reg_log_hta_smote.joblib")

print("Modelos de Regresión Logística guardados con éxito")
```

# ENTRENAMIENTO CON RANDOM FOREST

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' HIPERTENSIÓN 
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf_hta_class_weight = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_hta_class_weight.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_hta_class_weight, "../models/random_forest_hta_class_weight.joblib")
```

```python
# ==============================================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' Y 'sample_weight' HIPERTENSIÓN 
# ==============================================================================

rf_hta = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_hta.fit(X_train_hta, y_train_hta.values.ravel(), 
           sample_weight=sample_weight_train_hta.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_hta, "../models/random_forest_hta.joblib")
```

```python
# ======================================
# ENTRENAMIENTO RANDOM FOREST CON SMOTE 
# ======================================

smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())

print(f"Filas originales en train: {len(X_train_hta)} | Filas tras SMOTE: {len(X_train_smote)}")

rf_smote_hta = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1)

rf_smote_hta.fit(X_train_smote, y_train_smote)

print("MODELO RANDOM FOREST CON SMOTE ENTRENADO CORRECTAMENTE")

# Guardar modelo 
joblib.dump(rf_smote_hta, "../models/rf_smote_hta.joblib")
```

```python
# ========================================
# MODELO RANDOM FOREST C/ PESOS DINÁMICOS 
# ========================================

rf_subsample_hta = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample", # Pesos dinamicos por arbol 
    max_depth=10,  # controlar overfiting
    random_state=42,
    n_jobs=-1
)

rf_subsample_hta.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest con ajuste dinamico por árbol entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_subsample_hta, "../models/subsample_random_forest_hta.joblib")
```

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON Balanced Random Forest
# ============================================================

from imblearn.ensemble import BalancedRandomForestClassifier
brf_hta = BalancedRandomForestClassifier(
    n_estimators=300,
    sampling_strategy="all", # Equilibrar las clases en cada arbol
    replacement=True,
    random_state=42,
    n_jobs=-1
)

brf_hta.fit(X_train_hta, y_train_hta.values.ravel())

print("Modelo Random forest mejorado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_hta, "../models/balanced_random_forest_hta.joblib")
```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest con FACTORADULTO
# ===============================================


brf_hta_sample_weight = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=12, 
    min_samples_leaf=5,  
    sampling_strategy="auto",
    random_state=42,
    n_jobs=-1)

# Entrenar incorporando el peso del diseño muestral 
brf_hta_sample_weight.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel())


print("Modelo Balanced Random Forest con FACTORADULTO")

# Guardar modelo 
joblib.dump(brf_hta_sample_weight, "../models/brf_hta_sample_weight.joblib")
```

```python
# =================================================================
# OPTIMIZACIÓN: BALANCED RANDOM FOREST + GRIDSEARCH (HIPERTENSIÓN)
# =================================================================

from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

# Definimos el "menú" de hiperparámetros a probar
parametros_brf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2'],
    'class_weight': ['balanced', 'balanced_subsample'],
    'replacement': [False, True]
}

# Validación cruzada estratificada (mantiene la proporción de clases en cada fold, clave con datos desbalanceados)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Modelo base (sin ajustar todavía)
brf_hta_base = BalancedRandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

# cONFIGURAR RandomizedSearchCV
grid_search_brf = RandomizedSearchCV(
    estimator=brf_hta_base,
    param_distributions=parametros_brf,
    n_iter=15,               # Evalúa 15 combinaciones óptimas (solo 75 ajustados en total)
    scoring='roc_auc',
    cv=cv_estratificado,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

#  Entrenamos el GridSearch 
grid_search_brf.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel()
)

# Resultados de la búsqueda
print("Mejores hiperparámetros encontrados:")
print(grid_search_brf.best_params_)
print(f"\nMejor Recall promedio (validación cruzada): {grid_search_brf.best_score_:.4f}")

# Nos quedamos con el mejor modelo ya entrenado
brf_hta_optimo = grid_search_brf.best_estimator_

print("\nModelo Balanced Random Forest optimizado correctamente")


```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest optimizado
# ===============================================

brf_opt_gridsearch = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=4,
    max_features='sqrt',
    replacement=True,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

# Entrenar incorporando el peso del diseño muestral 
brf_opt_gridsearch.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel()
)

print("Modelo Balanced Random Forest optimizado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_opt_gridsearch, "../models/brf_hta_opt_gridsearch.joblib")
```

```python
# ===========================================================
# MODELO SVM 'class_weight' + 'sample_weight' - HIPERTENSIÓN 
# ===========================================================

from sklearn.svm import SVC

svm_hta = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

# Normalizo el peso 
sample_weight_norm = sample_weight_train_hta / sample_weight_train_hta.mean()

svm_hta.fit(X_train_hta, y_train_hta.values.ravel(), 
           sample_weight=sample_weight_norm.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_hta, "../models/svm_hta.joblib")
```

```python
# ==============================
# SVM CON SMOTE - HIPERTENSIÓN 
# ==============================
svm_smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_hta_smote, y_train_hta_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())

svm_hta_smote = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_hta_smote.fit(X_train_hta_smote, y_train_hta_smote)

print("SVM con SMOTE entrenado correctamente")

#Guardar modelo 
joblib.dump(svm_hta_smote, "../models/svm_hta_smote")
```

```python

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.svm import SVC

# Definir  hiperparámetros a optimizar (C y gamma)
param_svm = {
    "C": [0.1, 1.0, 5.0, 10.0],
    "gamma": ["scale", "auto", 0.01, 0.1],
}

# Configurar la Validación Cruzada Estratificada (5 Pliegues)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Instanciar el modelo base de SVM con class_weight='balanced'
svm_base = SVC(
    kernel="rbf",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42,
)

# Configurar GridSearchCV buscando maximizar roc_auc (o 'recall')
grid_search_svm = GridSearchCV(
    estimator=svm_base,
    param_grid=param_svm,
    scoring="roc_auc",  # Si prefieres optimizar directamente la Sensibilidad, usa scoring='recall'
    cv=cv_estratificado,
    verbose=2,
    n_jobs=-1,
)

# Ejecutar el ajuste fino (sin FACTORADULTO, solo con class_weight)
print(
    "Iniciando búsqueda de hiperparámetros para SVM (class_weight='balanced')..."
)
grid_search_svm.fit(X_train_hta, y_train_hta.values.ravel())

# Extraer el mejor modelo y sus parámetros
mejor_svm = grid_search_svm.best_estimator_

print("\n¡Ajuste fino de SVM completado!")
print(f"Mejor puntuación ROC-AUC en CV: {grid_search_svm.best_score_:.4f}")
print("Mejores Hiperparámetros encontrados:")
for param, val in grid_search_svm.best_params_.items():
    print(f"  - {param}: {val}")
```

```python
# =========================================================
# MODELO FINAL OPTIMIZADO SVM - HIPERTENSIÓN
# =========================================================

import joblib
from sklearn.svm import SVC

# Instanciar el modelo SVM con los hiperparámetros óptimos encontrados en CV
svm_hta_opt = SVC(
    kernel="rbf",
    C=1.0,
    gamma=0.01,
    class_weight="balanced",  # Gestión del desbalance
    probability=True,  # Necesario para calcular probabilidades y Curva ROC
    cache_size=1000,
    tol=1e-2,
    random_state=42,
)

# Entrenar el modelo final sobre todo el conjunto de entrenamiento
print("Entrenando modelo final optimizado de SVM...")
svm_hta_opt.fit(X_train_hta, y_train_hta.values.ravel())

print("¡Modelo SVM optimizado entrenado exitosamente!")

# Guardar el modelo 
joblib.dump(svm_hta_opt, "../models/svm_hta_optimizado.joblib")
```

# MODELOS XGBOOST

```python
# ================================================
# MODELO XGBOOST - 'class_weight' - HIPERTENSIÓN 
# ================================================
import joblib
from xgboost import XGBClassifier

# Gestión del desbalance equivalente a 'class_weight' 
y_train_limpio = y_train_hta.values.ravel()
equivalente_class_weight = float((y_train_limpio == 0).sum() / (y_train_limpio == 1).sum())

xgb_model_cw_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_cw_hta.fit(X_train_hta, y_train_limpio)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_cw_hta, "../models/xgb_model_cw_hta.joblib")
```

```python
# ==============================================================
# MODELO XGBOOST - 'class_weight' y FACTORADULTO - HIPERTENSIÓN 
# ==============================================================

xgb_model_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_hta.fit(X_train_hta, y_train_limpio, 
                  sample_weight=sample_weight_train_hta.values.ravel())

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_hta, "../models/xgb_model_hta.joblib")
```

```python
# ========================================
# MODELO XGBOOST - 'SMOTE' - HIPERTENSIÓN 
# ========================================

smote = SMOTE(random_state=42)
x_train_smote_xg, y_train_smote_xg = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel())


xgb_model_smote_hta = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    random_state = 42,
    n_jobs = -1)

xgb_model_smote_hta.fit(x_train_smote_xg, y_train_smote_xg)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_smote_hta, "../models/xgb_model_smote_hta.joblib")
```

```python
import numpy as np

# ¿Los dos modelos predicen probabilidades EXACTAMENTE iguales?
proba_cw = xgb_model_cw_hta.predict_proba(X_test_hta)[:, 1]
proba_factor = xgb_model_hta.predict_proba(X_test_hta)[:, 1]

diferencias = np.abs(proba_cw - proba_factor)

print(f"Diferencia máxima entre predicciones: {diferencias.max():.8f}")
print(f"Diferencia promedio: {diferencias.mean():.8f}")
print(f"Cantidad de predicciones IDÉNTICAS (diferencia = 0): {(diferencias == 0).sum()} de {len(diferencias)}")
```

```python
# ¿Cambiaron las CLASES predichas (no solo las probabilidades)?
pred_cw = xgb_model_cw_hta.predict(X_test_hta)
pred_factor = xgb_model_hta.predict(X_test_hta)

diferencias_clase = (pred_cw != pred_factor).sum()
print(f"Predicciones de clase que cambiaron: {diferencias_clase} de {len(pred_cw)}")

# Y veamos el AUC con más decimales (quizás sí difiere, solo que redondeado se veía igual)
from sklearn.metrics import roc_auc_score
auc_cw = roc_auc_score(y_test_hta, proba_cw)
auc_factor = roc_auc_score(y_test_hta, proba_factor)
print(f"AUC class_weight:  {auc_cw:.10f}")
print(f"AUC + FACTORADULTO: {auc_factor:.10f}")
```

```python
import joblib
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from xgboost import XGBClassifier

# Definir hiperparámetros a optimizar
param_xgb = {
    "n_estimators": [50, 100, 200],  # Número de árboles
    "max_depth": [3, 5, 7],  # Profundidad máxima del árbol
    "learning_rate": [0.01, 0.05, 0.1],  # Tasa de aprendizaje (eta)
    "subsample": [0.7, 0.8, 1.0],  # Proporción de submuestreo de filas
    "colsample_bytree": [
        0.7,
        0.8,
        1.0,
    ],  # Proporción de columnas en cada árbol
}

# Configurar la Validación Cruzada Estratificada (5 Pliegues)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Calcular la escala del desbalance
y_train_vec = y_train_hta.values.ravel()
ratio_desbalance = float((y_train_vec == 0).sum() / (y_train_vec == 1).sum())

# Instanciar el modelo base
xgb_base = XGBClassifier(
    scale_pos_weight=ratio_desbalance, random_state=42, n_jobs=-1
)

# Configurar GridSearchCV
grid_search_xgb = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_xgb,
    scoring="roc_auc",  # Métrica objetivo
    cv=cv_estratificado,
    verbose=1,
    n_jobs=-1,
)

# Ejecutar el ajuste fino (Fit) incorporando el FACTORADULTO
print("Iniciando búsqueda de hiperparámetros con Validación Cruzada...")
grid_search_xgb.fit(
    X_train_hta,
    y_train_vec,
    sample_weight=sample_weight_train_hta.values.ravel(),
)

# Extraer y mostrar el mejor modelo y sus parámetros
mejor_xgb = grid_search_xgb.best_estimator_
print("\n¡Ajuste fino completado!")
print(f"Mejor puntuación ROC AUCc en CV: {grid_search_xgb.best_score_:.4f}")
print("Mejores Hiperparámetros encontrados:")
for param, val in grid_search_xgb.best_params_.items():
    print(f"  - {param}: {val}")

```

```python
# =========================================================
# MODELO FINAL OPTIMIZADO XGBOOST - HIPERTENSIÓN
# =========================================================

import joblib
from xgboost import XGBClassifier

# Calcular el ratio de desbalance escalar
y_train_vec = y_train_hta.values.ravel()
ratio_desbalance = float(
    (y_train_vec == 0).sum() / (y_train_vec == 1).sum()
)

# Instanciar el modelo XGBoost con los hiperparámetros óptimos encontrados en CV
xgb_hta_opt = XGBClassifier(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.7,
    colsample_bytree=1.0,
    scale_pos_weight=ratio_desbalance,  # Balanceo de clases
    random_state=42,
    n_jobs=-1,
)

# Entrenar el modelo final sobre todo el conjunto de entrenamiento (incluyendo FACTORADULTO)
print("Entrenando modelo final optimizado de XGBoost...")
xgb_hta_opt.fit(
    X_train_hta,
    y_train_vec,
    sample_weight=sample_weight_train_hta.values.ravel(),
)

print("¡Modelo XGBoost optimizado entrenado exitosamente!")

# Guardar el modelo entrenado para la evaluación posterior
joblib.dump(xgb_hta_opt, "../models/xgb_hta_optimizado.joblib")
print(
    "Modelo guardado correctamente en: '../models/xgb_hta_optimizado.joblib'"
)
```
