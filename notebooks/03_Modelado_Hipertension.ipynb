```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# Cargar datos 
X_train_hta = pd.read_csv("../data/processed/X_train_hta_df.csv")
X_test_hta = pd.read_csv("../data/processed/X_test_hta_df.csv")
y_train_hta = pd.read_csv("../data/processed/y_train_hta.csv")
y_test_hta = pd.read_csv("../data/processed/y_test_hta.csv")

sample_weight_train_hta = pd.read_csv("../data/processed/sample_weight_train_hta.csv")

print("Datos listos para empezar a entrenar modelos")
```

# ENTRENAMIENTO CON REGRESIÓN LOGÍSTICA

```python
# ===================================================================
# ENTRENAMIENTO CON CLASS_WEIGHT Y FACTOR ADULTO CON AJUSTE DE UMBRAL 
# ===================================================================

reg_log_hta_class_weight = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Entrenar modelo
reg_log_hta_class_weight.fit(X_train_hta, y_train_hta.values.ravel())

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' ENTRENADA")
```

```python
# ======================================================
# ENTRENAMIENTO CON PESOS BALANCEADOS Y FACTOR ADULTO
# ======================================================

reg_log_hta = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Mismo código, pero añadiendo el sample_weight en el fit:
reg_log_hta.fit(
    X_train_hta,
    y_train_hta.values.ravel(),
    sample_weight=sample_weight_train_hta.values.ravel(),)

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' Y 'sample_weight' ENTRENADA")
```

```python
# ========================
# ENTRENAMIENTO CON SMOTE 
# ========================

from imblearn.over_sampling import SMOTE

# Se aplica SMOTE únicamente al conjunto de entrenamiento 
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_hta, y_train_hta.values.ravel()
)

# Se entrena la Regresión Logística con los datos sintéticos 
reg_log_hta_smote = LogisticRegression(max_iter=1000, random_state=42)
reg_log_hta_smote.fit(X_train_smote, y_train_smote)

print("MODELO REGRESIÓN LOGÍSTICA CON 'SMOTE' ENTRENADA")
```

```python
# =====================================
# GUARDAR MODELOS REGRESIÓN LOGÍSTICA 
# =====================================

import joblib

# Guardar modelo con 'class_weight'
joblib.dump(reg_log_hta_class_weight, "../models/reg_log_hta_class_weight.joblib")

# Guardar modelo con 'class_weight' y 'sample_weight' 
joblib.dump(reg_log_hta, "../models/reg_log_hta.joblib")

# Guardar modelo con SMOTE 
joblib.dump(reg_log_hta_smote, "../models/reg_log_hta_smote.joblib")

print("Modelos de Regresión Logística guardados con éxito")
```
