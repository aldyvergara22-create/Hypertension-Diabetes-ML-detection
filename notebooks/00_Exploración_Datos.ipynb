# DISTRIBUCIÓN DE LAS VARIABLES OBJETIVOS

```python
variables_objetivo = ["Hipertension_Diag_Medico", "Diabetes_Diag_Medico"]

#Bucle que analiza la frecuencia de las variables objetivo
for var in variables_objetivo:
    print(f"Variable: {var}")
    print(df_adulto[var].value_counts(dropna=False))
    print("\n")
    
    #Tabla de frecuencias
    conteo =  df_adulto[var].value_counts(dropna=False).sort_index()
    porcentaje = (df_adulto[var].value_counts(normalize=True, dropna=False).sort_index() * 100)
    
    #Modificación para evitar el KeyError: 'NaN' al intentar graficar las variables objetivo con valores nulos
    conteo.index = ["Nulos" if pd.isna(idx) else str(idx) for idx in conteo.index]
    porcentaje.index = conteo.index  # Asegura que los índices coincidan después de la modificación
    
    df_resumen = pd.DataFrame({"Conteo": conteo, "Porcentaje": porcentaje.round(2)})
    print(df_resumen)
    print("\n")
    
    df_resumen.to_csv(f"../reports/tables/Resumen_{var}.csv", index=True)
    
    #Graficos sinccronizados por color
    
    categorias = conteo.index.tolist()
    colores = sns.color_palette("viridis", n_colors=len(categorias))
    colores_dict = dict(zip(categorias, colores))
    
    #Gráficos 
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
    
    #Gráfico de barras
    sns.barplot(
        x=conteo.index,
        y=conteo.values,
        ax=axes[0],
        palette=colores_dict,
        hue=conteo.index,
        legend=False
    )
    axes[0].set_title(f"Distribución de {var} (Conteo)")
    axes[0].set_xlabel(var)
    axes[0].set_ylabel("Cantidad de registros")
    
    #Grafico de pastel sincronizado
    def formato_pct(pct):
        return f"{pct:.1f}%" if pct > 2.0 else ""
    
    etiquetas_limpias = [
        str(cat) if (val / conteo.sum() * 100) > 2.0 else ""
        for cat, val in zip(conteo.index, conteo.values)]

    axes[1].pie(
        conteo.values,
        labels=etiquetas_limpias,
        autopct=formato_pct,
        startangle=90,
        colors=[colores_dict[cat] for cat in conteo.index]
    )
    axes[1].set_title(f"Proporción de {var} (%)")
    
    plt.tight_layout()
    plt.savefig(f"../reports/figures/Distribucion_{var}.png", dpi=300, bbox_inches='tight')
    plt.show()
    
```

# ANÁLISIS DE LA VARIABLE FACTOR ADULTO

Comprobaciones principales:

- Sumatoria total: Debe coincidir aproximadamente con la población total de adultos del ambito estudiado.
- Ausencia de nulos o ceros: Ningun registro puede tener un factor negativo, cero, o nulo.
- Detección de sesgos: Se debe analizar la dispersión (mínimo, máximo, media) para comrpobar que no existan factores extremadamente de un solo encuestado. 

```python
col_factor = 'FACTORADULTO' 

print("--- AUDITORÍA DE LA VARIABLE FACTOR ---")

# Verificación de  nulos oficiales de Python (NaN)
nulos_pandas = df_adulto[col_factor].isna().sum()
porcentaje_nulos = (nulos_pandas / len(df_adulto)) * 100
print(f"Nulos oficiales (NaN): {nulos_pandas} ({porcentaje_nulos:.2f}%)")

# Verificación de valores en cero o negativos
ceros_o_negativos = (df_adulto[col_factor] <= 0).sum()
print(f"Valores menores o iguales a cero: {ceros_o_negativos}")

# Estadíticos descriptivos y estiamción de la población 
print("\n--- ESTADÍSTICOS DESCRIPTIVOS ---")
print(df_adulto[col_factor].describe())

población_estimada = df_adulto[col_factor].sum()
print(f"\nPoblación estimada según el factor de expansión: {población_estimada:,.0f}")

# VISUALIZACIÓN DE LA DISTRIBUCIÓN DEL FACTOR DE EXPANSIÓN
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Histograma del factor de expansión
sns.histplot(
    df_adulto[col_factor],
    kde=True,
    ax=axes[0],
    color=sns.color_palette("viridis")[2])

axes[0].set_title("Distribución del Factor de Expansión")
axes[0].set_xlabel("Peso poblacional (FACTORADULTO)")
axes[0].set_ylabel("Frecuencia")

# Boxplot del factor de expansión
sns.boxplot(
    x=df_adulto[col_factor],
    ax=axes[1],
    color=sns.color_palette("viridis")[4])

axes[1].set_title("Boxplot del Factor de Expansión")
axes[1].set_xlabel("Peso poblacional (FACTORADULTO)")

plt.tight_layout()
#plt.savefig("../reports/figures/Distribucion_Factor_Expansión.png", dpi=300, bbox_inches='tight')
plt.show()
```
