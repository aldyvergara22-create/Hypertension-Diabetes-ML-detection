# DISTRIBUCIÓN DE LAS VARIABLES OBJETIVOS (NO TERMINADO)

```python
variables_objetivo = ["Hipertension_Diag_Medico", "Diabetes_Diag_Medico"]

#Bucle que analiza la frecuencia de las variables objetivo
for var in variables_objetivo:
    print(f"Variable: {var}")
    print(df_adulto[var].value_counts(dropna=False))
    print("\n")
    
    #Tabla de frecuencias
    conteo =  df_adulto[var].value_counts(dropna=False)
    porcentaje = df_adulto[var].value_counts(normalize=True, dropna=False) * 100
    
    df_resumen = pd.DataFrame({"Conteo": conteo, "Porcentaje": porcentaje.round(2)})
    print(df_resumen)
    print("\n")
    
    #Gráficos 
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    #Gráfico de barras
    sns.countplot(
        data=df_adulto,
        x=var, 
        ax=axes[0],
        palette="viridis",
        hue=var,
        legend=False
    )
    axes[0].set_title(f"Distribución de {var} (Conteo)")
    axes[0].set_xlabel(var)
    axes[0].set_ylabel("Cantidad de registros")
    
    #Grafico de pastel
    datos_pie = df_adulto[var].value_counts(dropna=False)
    axes[1].pie(
        datos_pie,
        labels=datos_pie.index,
        autopct="%1.1f%%",
        startangle=90,
        colors=sns.color_palette("viridis", len(datos_pie))
    )
    axes[1].set_title(f"Proporción de {var} (%)")
    
    plt.tight_layout()
    plt.savefig(f"../reports/figures/Distribucion_{var}.png", dpi=300, bbox_inches='tight')
    plt.show()
    
```

# DISTRIBUCIÓN DE LAS VARIABLES NUMÉRICAS

```python
variables_numericas = ["Edad", "Hs_Sentado", "Cant_Diaria_Fruta", "Cant_Diaria_Verduras"]

for var in variables_numericas:
    print(f"Variable: {var}")
    print(df_adulto[var].value_counts(dropna=False))
    print("\n")
    
    #Guardo las estadísticas del describe en una variable
    estadisticas = (df_adulto[var].describe())
    print(estadisticas)
    print("\n")
    
    media = estadisticas["mean"]
    mediana = estadisticas["50%"]
    
    #Histogramas        
    plt.figure(figsize=(8, 5))
    sns.histplot(
        data=df_adulto,
        x=var,
        bins=18,
        kde=True,
        color="skyblue"
    )
    
    #Lineas de media y mediana extraidas de las estadísticas
    plt.axvline(media, color="red", linestyle="--", label=f"Media: {media:.2f}")
    plt.axvline(mediana, color="green", linestyle="--", label=f"Mediana: {mediana:.2f}")
    
    plt.title(f"Distribución de {var}")
    plt.xlabel(var)
    plt.ylabel("Frecuencia")
    plt.savefig(f"../reports/figures/Distribucion_de_{var}.png", dpi=300, bbox_inches='tight')
    plt.show()
```

## ANÁLISIS DE VARIABLES DE MEDICACIÓN Y CHEQUEOS

```python
# =================================================
# RELACIÓN DIAGNÓSTICO HIPERTENSION VS MEDICACIÓN 
# =================================================

crosstab_hipertension = (pd.crosstab(
    df_adulto["Hipertension_Diag_Medico"],
    df_adulto["Med_Hipertension_Cons"],
    margins=True,
    normalize="index"
) * 100).round(2)

# Se guarda la tabla en CSV 
crosstab_hipertension.to_csv("../reports/tables/crosstab_hipertension_vs_medicacion.csv", index=True)

print("Crosstab Hipertension vs Medicación (%):")
print(crosstab_hipertension)
```

```python
# ====================================================
# ANÁLISIS DE ADHERENCIA A MEDICACIÓN EN HIPERTENSOS
# ====================================================

#Filtro para pacientes con hipertensión diagnosticada
hipertensos = df_adulto[df_adulto["Hipertension_Diag_Medico"] == 1]

#Crosstab solo sobre la población con hipertensión diagnosticada
adherencia_hipertensos = (pd.crosstab(
    hipertensos["Med_Hipertension_Recetado"], 
    hipertensos["Med_Hipertension_Cons"], 
    rownames=["Prescrita por Médico"],
    colnames=["Consumida por el Paciente"],
    margins=True, 
    normalize="all"
    ) * 100
).round(2)

# Se guarda la tabla en CSV 
adherencia_hipertensos.to_csv("../reports/tables/adherencia_medicación_hipertensos.csv", index=True)

print("Crosstab Adherencia a Medicación en Hipertensos (%):")
print(adherencia_hipertensos)
```
