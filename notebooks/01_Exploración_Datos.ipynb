```python
def resumen_nulos(df, nombre):
    nulos = df.isnull().sum()
    pct   = (nulos / len(df) * 100).round(1)
    resumen = pd.DataFrame({"Nulos": nulos, "Porcentaje": pct})
    resumen = resumen[resumen["Nulos"] > 0].sort_values("Porcentaje", ascending=False)
    print(f"\n=== {nombre} — variables con nulos ===")
    print(resumen.head(30))

resumen_nulos(df_adulto, "ADULTO")
```

```python
# HIPERTENSIÓN → C5c_1  (diagnóstico médico confirmado)
# DIABETES     → C5c_12 (diagnóstico médico confirmado)

# Verificar distribución de cada una
for nombre, col in [("Hipertensión", "C5c_1"), ("Diabetes", "C5c_12")]:
    conteo = df_adulto[col].value_counts(dropna=False)
    total = len(df_adulto)
    print(f"\n=== {nombre} ({col}) ===")
    print(conteo)
    casos = (df_adulto[col] == 1).sum()
    print(f"Prevalencia: {casos/total*100:.1f}%")
```

```python
# La lógica correcta es:
# C5c_1 = 1  → diagnóstico confirmado por médico → POSITIVO
# C5c_1 = 2  → no diagnosticada                 → NEGATIVO  
# C5a_1 = 6  → nunca la ha padecido 
# → NEGATIVO
# NaN         → no aplicable (nunca la tuvo)     → NEGATIVO
# 9           → NS/NC                            → EXCLUIR

def construir_objetivo(df, col_a, col_c):
    objetivo = pd.Series(index=df.index, dtype='float64')
    # Positivos: diagnóstico médico confirmado
    objetivo[df[col_c] == 1] = 1
    # Negativos: nunca la tuvo O la tuvo pero no diagnosticada
    objetivo[(df[col_c] == 2) | (df[col_a] == 6)] = 0
    # NS/NC → NaN (los excluimos)
    objetivo[df[col_c] == 9] = pd.NA
    return objetivo

df_adulto['TARGET_HTA']      = construir_objetivo(df_adulto, 'C5a_1',  'C5c_1')
df_adulto['TARGET_DIABETES'] = construir_objetivo(df_adulto, 'C5a_12', 'C5c_12')

# Verificar el resultado
for nombre, col in [("Hipertensión", "TARGET_HTA"), ("Diabetes", "TARGET_DIABETES")]:
    conteo = df_adulto[col].value_counts(dropna=False)
    print(f"\n=== {nombre} ===")
    print(conteo)
    total_validos = df_adulto[col].notna().sum()
    positivos = (df_adulto[col] == 1).sum()
    print(f"Total válidos: {total_validos:,}")
    print(f"Prevalencia real: {positivos/total_validos*100:.1f}%")
```

```python
# CORRECCIÓN del target
def construir_objetivo_correcto(df, col_a, col_c):
    objetivo = pd.Series(0, index=df.index)  # todos empiezan como negativos
    
    # Positivos: diagnóstico médico confirmado
    objetivo[df[col_c] == 1] = 1
    
    # NS/NC en C5a → excluir (no sabemos)
    objetivo[df[col_a] == 9] = pd.NA
    
    # NS/NC en C5c → excluir
    objetivo[df[col_c] == 9] = pd.NA
    
    return objetivo

df_adulto['TARGET_HTA'] = construir_objetivo_correcto(
    df_adulto, 'C5a_1', 'C5c_1'
)
df_adulto['TARGET_DIABETES'] = construir_objetivo_correcto(
    df_adulto, 'C5a_12', 'C5c_12'
)

# Verificar
for nombre, col in [("Hipertensión", "TARGET_HTA"), ("Diabetes", "TARGET_DIABETES")]:
    total    = df_adulto[col].notna().sum()
    positivos = (df_adulto[col] == 1).sum()
    negativos = (df_adulto[col] == 0).sum()
    print(f"\n=== {nombre} ===")
    print(f"Positivos: {positivos:,}")
    print(f"Negativos: {negativos:,}")
    print(f"Total válidos: {total:,}")
    print(f"Prevalencia: {positivos/total*100:.1f}%")
```

```python
import numpy as np
import pandas as pd

# 1. Definimos el diccionario correcto (Código del INE -> Tu Nombre)
diccionario_nombres = {
    'IMC': 'IMC',
    'O2': 'Act. física',
    'O10_1': 'Hs. sentado',
    'Q1': 'Tabaquismo',
    'R1': 'Alcohol',
    'C1': 'Salud percib',
    'NIVEST': 'Nivel estudios',
    'P1_1': 'Alimentación'
}

# 2. Renombramos las columnas en tu DataFrame original
df_adulto.rename(columns=diccionario_nombres, inplace=True)

# 3. Creamos una copia con las variables ya renombradas para analizar los nulos
variables_renombradas = list(diccionario_nombres.values())
df_analisis = df_adulto[variables_renombradas].copy()

# 4. Mapeamos los códigos trampa por NaN reales de Python según el cuestionario
# Nota: La mayoría usa 9 (o 99), pero Hs. sentado al ser numérica suele usar 99 o 999.
df_analisis['IMC'] = df_analisis['IMC'].replace(9, np.nan)
df_analisis['Act. física'] = df_analisis['Act. física'].replace(9, np.nan)
df_analisis['Tabaquismo'] = df_analisis['Tabaquismo'].replace(9, np.nan)
df_analisis['Alcohol'] = df_analisis['Alcohol'].replace(9, np.nan)
df_analisis['Salud percib'] = df_analisis['Salud percib'].replace(9, np.nan)
df_analisis['Alimentación'] = df_analisis['Alimentación'].replace(9, np.nan)

# Códigos especiales para numéricas o clasificaciones largas:
df_analisis['Hs. sentado'] = df_analisis['Hs. sentado'].replace([99, 999], np.nan)
df_analisis['Nivel estudios'] = df_analisis['Nivel estudios'].replace([9, 99], np.nan)

# 5. Calculamos el porcentaje real de nulos ocultos
porcentaje_nulos = (df_analisis.isna().sum() / len(df_analisis)) * 100

print("--- PORCENTAJE DE VALORES NULOS OCULTOS POR VARIABLE ---")
print(porcentaje_nulos)
```
