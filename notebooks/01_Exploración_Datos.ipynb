# DISTRIBUCIÓN DE LAS VARIABLES OBJETIVOS

```python
variables_objetivo = ["Hipertension_Diag_Medico", "Diabetes_Diag_Medico"]

#Bucle que analiza la frecuencia de las variables objetivo
for var in variables_objetivo:
    print(f"Variable: {var}")
    print(df_adulto[var].value_counts(dropna=False))
    print("\n")
    
    #Tabla de frecuencias
    conteo =  df_adulto[var].value_counts(dropna=False).sort_index()
    porcentaje = (df_adulto[var].value_counts(normalize=True, dropna=False).sort_index() * 100)
    
    #Modificación para evitar el KeyError: 'NaN' al intentar graficar las variables objetivo con valores nulos
    conteo.index = ["Nulos" if pd.isna(idx) else str(idx) for idx in conteo.index]
    porcentaje.index = conteo.index  # Asegura que los índices coincidan después de la modificación
    
    df_resumen = pd.DataFrame({"Conteo": conteo, "Porcentaje": porcentaje.round(2)})
    print(df_resumen)
    print("\n")
    
    df_resumen.to_csv(f"../reports/tables/Resumen_{var}.csv", index=True)
    
    #Graficos sinccronizados por color
    
    categorias = conteo.index.tolist()
    colores = sns.color_palette("viridis", n_colors=len(categorias))
    colores_dict = dict(zip(categorias, colores))
    
    #Gráficos 
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
    
    #Gráfico de barras
    sns.barplot(
        x=conteo.index,
        y=conteo.values,
        ax=axes[0],
        palette=colores_dict,
        hue=conteo.index,
        legend=False
    )
    axes[0].set_title(f"Distribución de {var} (Conteo)")
    axes[0].set_xlabel(var)
    axes[0].set_ylabel("Cantidad de registros")
    
    #Grafico de pastel sincronizado
    def formato_pct(pct):
        return f"{pct:.1f}%" if pct > 2.0 else ""
    
    etiquetas_limpias = [
        str(cat) if (val / conteo.sum() * 100) > 2.0 else ""
        for cat, val in zip(conteo.index, conteo.values)]

    axes[1].pie(
        conteo.values,
        labels=etiquetas_limpias,
        autopct=formato_pct,
        startangle=90,
        colors=[colores_dict[cat] for cat in conteo.index]
    )
    axes[1].set_title(f"Proporción de {var} (%)")
    
    plt.tight_layout()
    plt.savefig(f"../reports/figures/Distribucion_{var}.png", dpi=300, bbox_inches='tight')
    plt.show()
    
```

## ANÁLISIS DE VARIABLES DE MEDICACIÓN Y CHEQUEOS DE HIPERTENSIÓN

```python
df_eda = df_adulto.copy()

#Hipertension
df_eda["Hipertension"] = (
    df_eda["Hipertension_Diag_Medico"].replace({
        1.0: "Sí",
        2.0: "No",
        9.0: "No contesta"
    })
)
df_eda.loc[
    df_eda["Hipertension_Diag_Medico"].isna(), "Hipertension"
] = "No"

#Diabetes
df_eda["Diabetes"] = (
    df_eda["Diabetes_Diag_Medico"].replace({
        1.0: "Sí",
        2.0: "No",
        9.0: "No contesta"
    })
)
df_eda.loc[
    df_eda["Diabetes_Diag_Medico"].isna(), "Diabetes"
] = "No"
```

```python
# =================================================
# RELACIÓN DIAGNÓSTICO HIPERTENSION VS MEDICACIÓN 
# =================================================

crosstab_hipertension = (pd.crosstab(
    df_adulto["Hipertension_Diag_Medico"],
    df_adulto["Med_Hipertension_Cons"],
    margins=True,
    normalize="index"
) * 100).round(2)

# Se guarda la tabla en CSV 
crosstab_hipertension.to_csv("../reports/tables/crosstab_hipertension_vs_medicacion.csv", index=True)

print("Crosstab Hipertension vs Medicación (%):")
print(crosstab_hipertension)
```

```python
# ====================================================
# ANÁLISIS DE ADHERENCIA A MEDICACIÓN EN HIPERTENSOS
# ====================================================

#Filtro para pacientes con hipertensión diagnosticada
hipertensos = df_adulto[df_adulto["Hipertension_Diag_Medico"] == 1].copy()

#Crosstab solo sobre la población con hipertensión diagnosticada
adherencia_hipertensos = (pd.crosstab(
    hipertensos["Med_Hipertension_Recetado"], 
    hipertensos["Med_Hipertension_Cons"], 
    rownames=["Prescrita por Médico"],
    colnames=["Consumida por el Paciente"],
    margins=True, 
    normalize="all"
    ) * 100
).round(2)

# Se guarda la tabla en CSV 
adherencia_hipertensos.to_csv("../reports/tables/adherencia_medicación_hipertensos.csv", index=True)

print("Crosstab Adherencia a Medicación en Hipertensos (%):")
print(adherencia_hipertensos)
```

```python
# ==================================================
# ANÁLISIS DE LA TEMPORALIDAD DE LA HIPERTENSIÓN
# ==================================================

crosstab_12m = (pd.crosstab(
    df_adulto["Hipertension_Diag_Medico"],
    df_adulto["Hipertension_Ultimos_12m"],
    margins=True,
    normalize="index"
) * 100).round(2)

crosstab_12m.to_csv("../reports/tables/Crosstab_Hipertension_12m.csv", index=True)

print("Crosstab Hipertension Diagnóstico vs Últimos 12 meses (%):")
print(crosstab_12m)
```

- Detección de sesgos: Se debe analizar la dispersión (mínimo, máximo, media) para comrpobar que no existan factores extremadamente de un solo encuestado.

### SITUACION DE LAS VARIBLES CON CÓDIGOS ESPECIALES y NaN


Al momento de escoger el tratamiento de estos codigos especiales se ha puesto como principal objetivo preservar al máximo los 2005 casos de diabetes y los 5774 de hipertension, para no aumentar el desbalance de clases. Para esto se tomarán las siguientes medidas:

- Se eliminará del modelo las variables Cant_Diaria_fruta y Cant_Diaria_Verduras. Esto se debe a que en ellas existen una cantidad de nulos importante y dado que tenemos las varaibles Cons_Fruta y Cons_Verduras incluir las primeras mencionadas pueden aportar poca información adicional y generar redundancia, a su vez se reduce la complejidad del modelo. 

- Variables categoricas con codigo especial: Para las variables Nivel_Estudios, Clase_social,
Act_Fisica, Tabaquismo, Cons_Alcohol se optará por usar One-Hot_Encoding para tener una columna como "Tabaquismo_No_Contesta", y que sea el modelo quien decida si esa categoria aporta información o no. 

- Variables de frecuencia de consumo alimenticio: imputar la moda en las variables de frecuencia de consumo alimentario, ya que el porcentaje de faltantes es muy pequeño y "No contesta" no representa un comportamiento alimentario.

- Variables númericas: IMC y Hs_Sentado se van a imputar con la mediana.

- Variables objetivo: Esta es la unica excepción donde si eliminamos registros, 3 de diabetes y 31 de hipertensos que no contestaron y los NaN dado la naturaleza de la encuesta son negativos por lo que debemos asignarlos como negativos con el codigo 2.

# CORRELACIONES ENTRE VARIABLES NUMÉRICAS

```python
var_numericas = ["Edad", "Hs_Sentado"]

# Copia temporal que reemplaza códigos especiales por NaN para no sesgar la correlación
df_corr = df_adulto[var_numericas].copy()
df_corr["Hs_Sentado"] = df_corr["Hs_Sentado"].replace([99, 99.0], np.nan)

# Calculo de la matriz de correlación de Pearson
matriz_corr = df_corr.corr(method="pearson")

# Gráfico Heatmap
plt.figure(figsize=(5, 4))
sns.heatmap(
    matriz_corr,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    cbar=True,
    linewidths=1
)

plt.title("Matriz de Correlación: Edad vs. Horas Sentado", fontsize=11)
plt.tight_layout()
plt.savefig("../reports/figures/matriz_correlacion_edad_HsSentado.png", dpi=300, bbox_inches="tight")
plt.show()

# Imprimir el valor exacto del coeficiente
r = matriz_corr.loc["Edad", "Hs_Sentado"]
print(f"Coeficiente de correlación de Pearson (r): {r:.4f}")

```

### INTERPRETACION DE LA CORRELACIÓN:

El coeficiente de correlación de Pearson entre "Edad" y "Hs_Sentado" es de (r) = 0.0995
Este valor cercado al cero demuestra que no  existe correlación lineal significativa entre la 
edad del encuestado y la cantidad de horas diarias que permanece sentado. 
Ambos predictores aportan información independiente al análisis, descartando problemas de 
multicolinealidad.

# PRUEBA CHI CUADRADO DE INDEPENDENCIA

```python
# =============================================
# ANÁLISIS CHI-CUADRADO CONTRA HIPERTENSIÓN
# =============================================


from scipy.stats import chi2_contingency

def cramers_v(contingency_table):
    """Calcula el coeficiente V de Cramér para medir la fuerza de asociación."""
    chi2 = chi2_contingency(contingency_table)[0]
    n = contingency_table.sum().sum()
    phi2 = chi2 / n
    r, k = contingency_table.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1)) / (n-1))
    rcorr = r - ((r-1)**2) / (n-1)
    kcorr = k - ((k-1)**2) / (n-1)
    
    # Previene división por cero
    denom = min((kcorr-1), (rcorr-1))
    if denom == 0:
        return 0.0
    return np.sqrt(phi2corr / denom)

#  Defino lista de variables categóricas explicativas a evaluar contra Hipertensión
variables_categoricas = [
    'Sexo','Edad', 'Nivel_Estudios', 'Clase_Social',"IMC", 
    'Act_Fisica', 'Tabaquismo', 'Ex_Fumador', 'Cons_Alcohol',
    'Cons_Fruta', 'Cons_Carne', 'Cons_Huevo', 'Cons_Pescado', 
    'Cons_Carbohidrato', 'Cons_Cereales', 'Cons_Verduras', 
    'Cons_Legumbres', 'Cons_Embutidos', 'Cons_Lacteos', 
    'Cons_Dulces', 'Cons_Refrescos', 'Cons_Comida_Rapida', 
    'Cons_Snacks_Salados', 'Cons_Zumos_Naturales', 'Diabetes_Diag_Medico'
]

target_hipertension = 'Hipertension_Diag_Medico'

#  Itera y calcula Chi2, p-valor y V de Cramér
resultados = []

for var in variables_categoricas:
    if var in df_adulto.columns and target_hipertension in df_adulto.columns:
        # Tabla de contingencia omitiendo nulos
        tabla_contingencia = pd.crosstab(df_adulto[var], df_adulto[target_hipertension])
        
        # Prueba de Chi-cuadrado
        chi2, p, dof, _ = chi2_contingency(tabla_contingencia)
        
        # V de Cramér
        v_cramer = cramers_v(tabla_contingencia)
        
        # Guardar en la lista
        resultados.append({
            'Variable': var,
            'Chi2': round(chi2, 4),
            'p-valor': p,
            'Significativo (p < 0.05)': 'Sí' if p < 0.05 else 'No',
            'V de Cramér': round(v_cramer, 4)
        })

# Se convierten resultados a DataFrame y se ordenan por V de Cramér (mayor a menor)
df_chi2_hipertension = pd.DataFrame(resultados).sort_values(by='V de Cramér', ascending=False)

#guardar en CSV
df_chi2_hipertension.to_csv("../reports/tables/Análisis_CHI_CUADRADO.csv", index=True)

# Mostrar tabla ordenada
print("=== ANÁLISIS CHI-CUADRADO CONTRA HIPERTENSIÓN DIAGNÓSTICO MÉDICO ===")
display(df_chi2_hipertension)
```

### Conclusión resultados

Todas las variables evaluadas —excepto Sexo— presentan una relación estadísticamente significativa con la hipertensión (p < 0.05). Aunque las fuerzas de asociación según el V de Cramér son modestas (V < 0.10), destacan los factores dietéticos (snacks salados, comida rápida) y el nivel educativo como los predictores categóricos de mayor peso relativo
