```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# Cargar datos 

X_train_diab = pd.read_csv("../data/processed/X_train_diab_df.csv")
X_test_diab = pd.read_csv("../data/processed/X_test_diab_df.csv")
y_train_diab = pd.read_csv("../data/processed/y_train_diab.csv")
y_test_diab = pd.read_csv("../data/processed/y_test_diab.csv")

sample_weight_train_diab =  pd.read_csv("../data/processed/sample_weight_train_diab.csv")

print("Datos listos para empezar a entrenar modelos")
```

# ENTRENAMIENTO CON REGRESIÓN LOGÍSTICA

```python
# ===================================================================
# ENTRENAMIENTO CON CLASS_WEIGHT Y FACTOR ADULTO CON AJUSTE DE UMBRAL 
# ===================================================================

reg_log_diab_class_weight = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Entrenar modelo
reg_log_diab_class_weight.fit(X_train_diab, y_train_diab.values.ravel())

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' ENTRENADA")
```

```python
# ======================================================
# ENTRENAMIENTO CON PESOS BALANCEADOS Y FACTOR ADULTO
# ======================================================

reg_log_diab = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Mismo código, pero añadiendo el sample_weight en el fit:
reg_log_diab.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel(),)

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' Y 'sample_weight' ENTRENADA")
```

```python
# ========================
# ENTRENAMIENTO CON SMOTE 
# ========================

from imblearn.over_sampling import SMOTE

# Se aplica SMOTE únicamente al conjunto de entrenamiento 
smote = SMOTE(random_state=42)
X_train_diab_smote, y_train_diab_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel()
)

# Se entrena la Regresión Logística con los datos sintéticos 
reg_log_diab_smote = LogisticRegression(max_iter=1000, random_state=42)
reg_log_diab_smote.fit(X_train_diab_smote, y_train_diab_smote)

print("MODELO REGRESIÓN LOGÍSTICA CON 'SMOTE' ENTRENADA")
```

```python
# =====================================
# GUARDAR MODELOS REGRESIÓN LOGÍSTICA 
# =====================================

import joblib

# Guardar modelo con 'class_weight'
joblib.dump(reg_log_diab_class_weight, "../models/reg_log_diab_class_weight.joblib")

# Guardar modelo con 'class_weight' y 'sample_weight' 
joblib.dump(reg_log_diab, "../models/reg_log_diab.joblib")

# Guardar modelo con SMOTE 
joblib.dump(reg_log_diab_smote, "../models/reg_log_diab_smote.joblib")

print("Modelos de Regresión Logística guardados con éxito")
```

# ENTRENAMIENTO RANDOM FOREST

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' DIABETES 
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf_diab_class_weight = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_diab_class_weight.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_diab_class_weight, "../models/random_forest_diab_class_weight.joblib")
```

```python
# =================================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' Y 'sample_weight' 
# =================================================================

rf_diab = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_diab.fit(X_train_diab, y_train_diab.values.ravel(), 
           sample_weight=sample_weight_train_diab.values.ravel())

print("Modelo Random forest class_weight y sample_weight entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_diab, "../models/random_forest_diab.joblib")
```

```python
# ======================================
# ENTRENAMIENTO RANDOM FOREST CON SMOTE 
# ======================================

smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel())

print(f"Filas originales en train: {len(X_train_diab)} | Filas tras SMOTE: {len(X_train_smote)}")

rf_smote_diab = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1)

rf_smote_diab.fit(X_train_smote, y_train_smote)

print("MODELO RANDOM FOREST CON SMOTE ENTRENADO CORRECTAMENTE")

# Guardar modelo 
joblib.dump(rf_smote_diab, "../models/rf_smote_diabetes.joblib")
```

```python
# ========================================
# MODELO RANDOM FOREST C/ PESOS DINÁMICOS 
# ========================================

rf_subsample = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample", # Pesos dinamicos por arbol 
    max_depth=10,  # controlar overfiting
    random_state=42,
    n_jobs=-1
)

rf_subsample.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest con ajuste dinamico por árbol entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_subsample, "../models/subsample_random_forest_diab.joblib")
```

```python
# ============================================
# ENTRENAMIENTO  Balanced Random Forest 'class_weight'
# =============================================

from imblearn.ensemble import BalancedRandomForestClassifier
brf_diabetes = BalancedRandomForestClassifier(
    n_estimators=300,
    sampling_strategy="all", # Equilibrar las clases en cada arbol
    replacement=True,
    random_state=42,
    n_jobs=-1
)

brf_diabetes.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest mejorado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_diabetes, "../models/balanced_random_forest_diab.joblib")
```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest con FACTORADULTO
# ===============================================
from imblearn.ensemble import BalancedRandomForestClassifier
import joblib

brf_sample_weight_diab = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=12, 
    min_samples_leaf=5,  
    sampling_strategy="auto",
    random_state=42,
    n_jobs=-1)

# Entrenar incorporando el peso del diseño muestral 
brf_sample_weight_diab.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel())


print("Modelo Balanced Random Forest optimizado")

# Guardar modelo 
joblib.dump(brf_sample_weight_diab, "../models/brf_sample_weight_diab.joblib")
```

```python
# =================================================================
# OPTIMIZACIÓN: BALANCED RANDOM FOREST + GRIDSEARCH (DIABETES)
# =================================================================

from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

# Definimos el "menú" de hiperparámetros a probar
parametros_brf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2'],
    'class_weight': ['balanced', 'balanced_subsample'],
    'replacement': [False, True]
}
# Validación cruzada estratificada (mantiene la proporción de clases en cada fold, clave con datos desbalanceados)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Modelo base (sin ajustar todavía)
brf_diab_base = BalancedRandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

# GridSearch: prueba todas las combinaciones y se queda con la que mejor Recall promedio da en los 5 folds
grid_search_brf = RandomizedSearchCV(
    estimator=brf_diab_base,
    param_distributions=parametros_brf,
    n_iter=15,               # Evalúa 15 combinaciones óptimas (solo 75 ajustados en total)
    scoring='roc_auc',
    cv=cv_estratificado,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

#  Entrenamos el GridSearch 
grid_search_brf.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel()
)

# Resultados de la búsqueda
print("Mejores hiperparámetros encontrados:")
print(grid_search_brf.best_params_)
print(f"\nMejor ROC_AUC promedio (validación cruzada): {grid_search_brf.best_score_:.4f}")

# Nos quedamos con el mejor modelo ya entrenado
brf_diab_optimo = grid_search_brf.best_estimator_

print("\nModelo Balanced Random Forest optimizado correctamente")


```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest optimizado
# ===============================================


brf_opt_gridsearch = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=2,
    max_features='sqrt',
    replacement=False,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

# Entrenar incorporando el peso del diseño muestral 
brf_opt_gridsearch.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel()
)

print("Modelo Balanced Random Forest optimizado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_opt_gridsearch, "../models/brf_hta_opt_gridsearch.joblib")
```

```python
# =========================================
# MODELO SVM 'class_weight' - DIABETES 
# =========================================

from sklearn.svm import SVC

svm_diab_cw = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

# Entrenamiento 
svm_diab_cw.fit(X_train_diab, y_train_diab.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_diab_cw, "../models/svm_hta_diab.joblib")
```

```python
# ===========================================================
# MODELO SVM 'class_weight' + 'sample_weight' - DIABETES 
# ===========================================================

import joblib
from sklearn.svm import SVC

svm_diab = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

# Normalizo el peso 
sample_weight_norm = sample_weight_train_diab / sample_weight_train_diab.mean()

svm_diab.fit(X_train_diab, y_train_diab.values.ravel(), 
           sample_weight=sample_weight_norm.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_diab, "../models/svm_diab.joblib")

```

```python
# ==========================
# SVM CON SMOTE - DIABETES
# ==========================

svm_smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_diab_smote, y_train_diab_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel())

svm_diab_smote = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_diab_smote.fit(X_train_diab_smote, y_train_diab_smote)

print("SVM con SMOTE entrenado correctamente")

#Guardar modelo 
joblib.dump(svm_diab_smote, "../models/svm_diab_smote.joblib")
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.svm import SVC


# Definir  hiperparámetros a optimizar (C y gamma)
param_svm = {
    "C": [0.1, 1.0, 5.0, 10.0],
    "gamma": ["scale", "auto", 0.01, 0.1],
}

# Configurar la Validación Cruzada Estratificada (5 Pliegues)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Instanciar el modelo base de SVM con class_weight='balanced'
svm_base_diab = SVC(
    kernel="rbf",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42,
)

# Configurar RandomizedSearchCV
random_search_svm_diab = RandomizedSearchCV(
    estimator=svm_base_diab,
    param_distributions=param_svm,
    n_iter=6,  # Limita a solo 6 combinaciones para máxima rapidez
    scoring="roc_auc",
    cv=cv_estratificado,
    verbose=2,
    random_state=42,
    n_jobs=-1,
)

# Ejecutar el ajuste fino de diabetes
print("Iniciando búsqueda rápida (RandomizedSearchCV) para SVM en Diabetes...")
random_search_svm_diab.fit(X_train_diab, y_train_diab.values.ravel())

# Extraer el mejor modelo y sus parámetros
mejor_svm_diab = random_search_svm_diab.best_estimator_

print("\n¡Ajuste fino de SVM completado!")
print(f"Mejor puntuación ROC-AUC en CV: {random_search_svm_diab.best_score_:.4f}")
print("Mejores Hiperparámetros encontrados:")
for param, val in random_search_svm_diab.best_params_.items():
    print(f"  - {param}: {val}")
```

```python
# ========================================
# MODELO FINAL OPTIMIZADO SVM - DIABETES
# ========================================

import joblib
from sklearn.svm import SVC

# Instanciar el modelo SVM con los hiperparámetros óptimos encontrados en CV
svm_diab_opt = SVC(
    kernel="rbf",
    C=0.1,
    gamma="auto",
    class_weight="balanced",  # Gestión del desbalance
    probability=True,  # Necesario para calcular probabilidades y Curva ROC
    cache_size=1000,
    tol=1e-2,
    random_state=42,
)

# Entrenar el modelo final sobre todo el conjunto de entrenamiento
print("Entrenando modelo final optimizado de SVM...")
svm_diab_opt.fit(X_train_diab, y_train_diab.values.ravel())

print("¡Modelo SVM optimizado entrenado exitosamente!")

# Guardar el modelo 
joblib.dump(svm_diab_opt, "../models/svm_diab_optimizado.joblib")
```

# MODELOS XGBOOST

```python
# ===========================================
# MODELO XGBOOST - 'class_weight' - DIABETES
# ===========================================

from xgboost import XGBClassifier

# Gestión del desbalance equivalente a 'class_weight' 
y_train_limpio = y_train_diab.values.ravel()
equivalente_class_weight = float((y_train_limpio == 0).sum() / (y_train_limpio == 1).sum())

xgb_model_cw_diab = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_cw_diab.fit(X_train_diab, y_train_limpio)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_cw_diab, "../models/xgb_model_cw_diab.joblib")
```

```python
# ==========================================================
# MODELO XGBOOST - 'class_weight' y FACTORADULTO - DIABETES
# ==========================================================

xgb_model_diab = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    scale_pos_weight = equivalente_class_weight,
    random_state = 42,
    n_jobs = -1)

xgb_model_diab.fit(X_train_diab, y_train_limpio, 
                  sample_weight=sample_weight_train_diab.values.ravel())

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_diab, "../models/xgb_model_diab.joblib")
```

```python
# ====================================
# MODELO XGBOOST - 'SMOTE' - DIABETES 
# ====================================

smote = SMOTE(random_state=42)
x_train_smote_xg, y_train_smote_xg = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel())


xgb_model_smote_diab = XGBClassifier(
    n_estimators = 100,
    learning_rate = 0.1,
    max_depth = 6,
    random_state = 42,
    n_jobs = -1)

xgb_model_smote_diab.fit(x_train_smote_xg, y_train_smote_xg)

print("XGBoost entrenado correctamente.")

# Guardar modelo 
joblib.dump(xgb_model_smote_diab, "../models/xgb_model_smote_diab.joblib")
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from xgboost import XGBClassifier

# Definir hiperparámetros a optimizar
param_xgb = {
    "n_estimators": [50, 100, 200],  # Número de árboles
    "max_depth": [3, 5, 7],  # Profundidad máxima del árbol
    "learning_rate": [0.01, 0.05, 0.1],  # Tasa de aprendizaje (eta)
    "subsample": [0.7, 0.8, 1.0],  # Proporción de submuestreo de filas
    "colsample_bytree": [
        0.7,
        0.8,
        1.0,
    ],  # Proporción de columnas en cada árbol
}

# Configurar la Validación Cruzada Estratificada (5 Pliegues)
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Calcular la escala del desbalance
y_train_vec = y_train_diab.values.ravel()
ratio_desbalance = float((y_train_vec == 0).sum() / (y_train_vec == 1).sum())

# Instanciar el modelo base
xgb_base = XGBClassifier(
    scale_pos_weight=ratio_desbalance, random_state=42, n_jobs=-1
)

# Configurar GridSearchCV
grid_search_xgb = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_xgb,
    scoring="roc_auc",  # Métrica objetivo
    cv=cv_estratificado,
    verbose=1,
    n_jobs=-1,
)

# Ejecutar el ajuste fino (Fit) incorporando el FACTORADULTO
print("Iniciando búsqueda de hiperparámetros con Validación Cruzada...")
grid_search_xgb.fit(
    X_train_diab,
    y_train_vec,
    sample_weight=sample_weight_train_diab.values.ravel(),
)

# Extraer y mostrar el mejor modelo y sus parámetros
mejor_xgb = grid_search_xgb.best_estimator_
print("\n¡Ajuste fino completado!")
print(f"Mejor puntuación ROC AUCc en CV: {grid_search_xgb.best_score_:.4f}")
print("Mejores Hiperparámetros encontrados:")
for param, val in grid_search_xgb.best_params_.items():
    print(f"  - {param}: {val}")

```

```python
# ============================================
# MODELO FINAL OPTIMIZADO XGBOOST - DIABETES
# ============================================

import joblib
from xgboost import XGBClassifier

# Calcular el ratio de desbalance escalar
y_train_vec = y_train_diab.values.ravel()
ratio_desbalance = float(
    (y_train_vec == 0).sum() / (y_train_vec == 1).sum()
)

# Instanciar el modelo XGBoost con los hiperparámetros óptimos encontrados en CV
xgb_hta_opt = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.7,
    colsample_bytree=1.0,
    scale_pos_weight=ratio_desbalance,  # Balanceo de clases
    random_state=42,
    n_jobs=-1,
)

# Entrenar el modelo final sobre todo el conjunto de entrenamiento (incluyendo FACTORADULTO)
print("Entrenando modelo final optimizado de XGBoost...")
xgb_hta_opt.fit(
    X_train_diab,
    y_train_vec,
    sample_weight=sample_weight_train_diab.values.ravel(),
)

print("¡Modelo XGBoost optimizado entrenado exitosamente!")

# Guardar el modelo entrenado para la evaluación posterior
joblib.dump(xgb_hta_opt, "../models/xgb_diab_optimizado.joblib")
```
