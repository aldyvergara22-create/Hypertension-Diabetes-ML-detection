```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# Cargar datos 

X_train_diab = pd.read_csv("../data/processed/X_train_diab_df.csv")
X_test_diab = pd.read_csv("../data/processed/X_test_diab_df.csv")
y_train_diab = pd.read_csv("../data/processed/y_train_diab.csv")
y_test_diab = pd.read_csv("../data/processed/y_test_diab.csv")

sample_weight_train_diab =  pd.read_csv("../data/processed/sample_weight_train_diab.csv")

print("Datos listos para empezar a entrenar modelos")
```

# ENTRENAMIENTO CON REGRESIÓN LOGÍSTICA

```python
# ===================================================================
# ENTRENAMIENTO CON CLASS_WEIGHT Y FACTOR ADULTO CON AJUSTE DE UMBRAL 
# ===================================================================

reg_log_diab_class_weight = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Entrenar modelo
reg_log_diab_class_weight.fit(X_train_diab, y_train_diab.values.ravel())

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' ENTRENADA")
```

```python
# ======================================================
# ENTRENAMIENTO CON PESOS BALANCEADOS Y FACTOR ADULTO
# ======================================================

reg_log_diab = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)

# Mismo código, pero añadiendo el sample_weight en el fit:
reg_log_diab.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel(),)

print("MODELO REGRESIÓN LOGÍSTICA CON 'class_weight' Y 'sample_weight' ENTRENADA")
```

```python
# ========================
# ENTRENAMIENTO CON SMOTE 
# ========================

from imblearn.over_sampling import SMOTE

# Se aplica SMOTE únicamente al conjunto de entrenamiento 
smote = SMOTE(random_state=42)
X_train_diab_smote, y_train_diab_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel()
)

# Se entrena la Regresión Logística con los datos sintéticos 
reg_log_diab_smote = LogisticRegression(max_iter=1000, random_state=42)
reg_log_diab_smote.fit(X_train_diab_smote, y_train_diab_smote)

print("MODELO REGRESIÓN LOGÍSTICA CON 'SMOTE' ENTRENADA")
```

```python
# =====================================
# GUARDAR MODELOS REGRESIÓN LOGÍSTICA 
# =====================================

import joblib

# Guardar modelo con 'class_weight'
joblib.dump(reg_log_diab_class_weight, "../models/reg_log_diab_class_weight.joblib")

# Guardar modelo con 'class_weight' y 'sample_weight' 
joblib.dump(reg_log_diab, "../models/reg_log_diab.joblib")

# Guardar modelo con SMOTE 
joblib.dump(reg_log_diab_smote, "../models/reg_log_diab_smote.joblib")

print("Modelos de Regresión Logística guardados con éxito")
```

# ENTRENAMIENTO RANDOM FOREST

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' DIABETES 
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf_diab_class_weight = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_diab_class_weight.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_diab_class_weight, "../models/random_forest_diab_class_weight.joblib")
```

```python
# =================================================================
# ENTRENAMIENTO RANDOM FOREST CON 'class_weight' Y 'sample_weight' 
# =================================================================

rf_diab = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1)

rf_diab.fit(X_train_diab, y_train_diab.values.ravel(), 
           sample_weight=sample_weight_train_diab.values.ravel())

print("Modelo Random forest class_weight y sample_weight entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_diab, "../models/random_forest_diab.joblib")
```

```python
# ======================================
# ENTRENAMIENTO RANDOM FOREST CON SMOTE 
# ======================================

smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel())

print(f"Filas originales en train: {len(X_train_diab)} | Filas tras SMOTE: {len(X_train_smote)}")

rf_smote_diab = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1)

rf_smote_diab.fit(X_train_smote, y_train_smote)

print("MODELO RANDOM FOREST CON SMOTE ENTRENADO CORRECTAMENTE")

# Guardar modelo 
joblib.dump(rf_smote_diab, "../models/rf_smote_diabetes.joblib")
```

```python
# ========================================
# MODELO RANDOM FOREST C/ PESOS DINÁMICOS 
# ========================================

rf_subsample = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample", # Pesos dinamicos por arbol 
    max_depth=10,  # controlar overfiting
    random_state=42,
    n_jobs=-1
)

rf_subsample.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest con ajuste dinamico por árbol entrenado correctamente")

# Guardar modelo 
joblib.dump(rf_subsample, "../models/subsample_random_forest_diab.joblib")
```

```python
# ============================================================
# ENTRENAMIENTO RANDOM FOREST CON Balanced Random Forest
# ============================================================

from imblearn.ensemble import BalancedRandomForestClassifier
brf_diabetes = BalancedRandomForestClassifier(
    n_estimators=300,
    sampling_strategy="all", # Equilibrar las clases en cada arbol
    replacement=True,
    random_state=42,
    n_jobs=-1
)

brf_diabetes.fit(X_train_diab, y_train_diab.values.ravel())

print("Modelo Random forest mejorado entrenado correctamente")

# Guardar modelo 
joblib.dump(brf_diabetes, "../models/balanced_random_forest_diab.joblib")
```

```python
# ===============================================
# ENTRENAMIENTO Balanced Random Forest optimizado
# ===============================================


brf_opt = BalancedRandomForestClassifier(
    n_estimators=300,
    max_depth=12, 
    min_samples_leaf=5,  
    sampling_strategy="auto",
    random_state=42,
    n_jobs=-1)

# Entrenar incorporando el peso del diseño muestral 
brf_opt.fit(
    X_train_diab,
    y_train_diab.values.ravel(),
    sample_weight=sample_weight_train_diab.values.ravel())


print("Modelo Balanced Random Forest optimizado")

# Guardar modelo 
joblib.dump(brf_opt, "../models/brf_diab_opt.joblib")
```

# SVM

```python
# =========================================
# MODELO SVM 'class_weight' - DIABETES 
# =========================================

from sklearn.svm import SVC

svm_diab_cw = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

# Entrenamiento 
svm_diab_cw.fit(X_train_diab, y_train_diab.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_diab_cw, "../models/svm_hta_diab.joblib")
```

```python
# ===========================================================
# MODELO SVM 'class_weight' + 'sample_weight' - DIABETES 
# ===========================================================


svm_diab = SVC(
    kernel= "rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_diab.fit(X_train_diab, y_train_diab.values.ravel(), 
           sample_weight=sample_weight_train_diab.values.ravel())

print("SVM entrenado correctamente.")

# Guardar modelo 
joblib.dump(svm_diab, "../models/svm_diab.joblib")
```

```python
# ==========================
# SVM CON SMOTE - DIABETES
# ==========================

svm_smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_diab_smote, y_train_diab_smote = smote.fit_resample(
    X_train_diab, y_train_diab.values.ravel())

svm_diab_smote = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    probability=True,
    cache_size=1000,
    tol=1e-2,
    random_state=42)

svm_diab_smote.fit(X_train_diab_smote, y_train_diab_smote)

print("SVM con SMOTE entrenado correctamente")

#Guardar modelo 
joblib.dump(svm_diab_smote, "../models/svm_diab_smote")
```
